```python
def LGBMModel(action, X_train, y_train, X_test, y_test, params):
    model = lgbm.LGBMClassifier(
              n_estimators=params['n_estimators'],
              boosting_type=params['boosting_type'],
              max_depth=params['max_depth'],
              num_leaves=params['num_leaves'],
              max_bin=params['max_bin'],
              learning_rate=params['learning_rate'],
              reg_alpha=params['reg_alpha'],
              reg_lambda=params['reg_lambda'],
              num_class=params['num_class'],
              device_type=params['device_type'],
              objective=params['objective'],
              tree_learner=params['tree_learner'],
              num_iterations=params['num_iterations'],
              eval_metric=params['eval_metric'])

    if action == 'init':
        return model
    elif action == 'fit':
        model.fit(X_train, y_train,
                            eval_set=[(X_test, y_test)],
                            eval_metric="logloss",
                            callbacks=[lgbm.log_evaluation(period=10),
                                        lgbm.early_stopping(stopping_rounds=10)]
                  )
        return model
```

```python
def get_predictions(y_prob, classes, threshold, top_n, y_true, le):
    y_pred = []

    keys = ['true_list', "pred_code1", "score1", "pred_code2", "score2", "pred_code3", "score3", "pred_code4", "score4", "pred_code5", "score5"]
    data_top_5 = {key: [] for key in keys}

    for idx, prob in enumerate(y_prob):
        above_threshold_indices = np.where(prob > threshold[1])[0]
        if len(above_threshold_indices) == 0:
            # If no classes are above the threshold, use the highest scores
            sorted_indices = np.argsort(-prob)[:top_n]
        else:
            # Sort filtered classes by probability in descending order
            above_threshold_probabilities = prob[above_threshold_indices]
            sorted_indices = above_threshold_indices[np.argsort(-above_threshold_probabilities)][:top_n]

        recommended_icd_code = [classes[i] for i in sorted_indices]
        recommended_score = [prob[i] for i in sorted_indices]
        if y_true[idx] in recommended_icd_code:
            y_pred.append(y_true[idx])
        else:
            y_pred.append(recommended_icd_code[0])

        # Ensure we have 5 elements in the lists
        while len(recommended_icd_code) < 5:
            recommended_icd_code.append(None)
            recommended_score.append(None)
        try:
            data_top_5['true_list'].append(le.inverse_transform([y_true[idx]])[0])
        except:
            data_top_5['true_list'].append(y_true[idx])

        try:
            data_top_5['pred_code1'].append(le.inverse_transform([recommended_icd_code[0]])[0])
        except:
            data_top_5['pred_code1'].append(recommended_icd_code[0])
        data_top_5['score1'].append(recommended_score[0])

        try:
            data_top_5['pred_code2'].append(le.inverse_transform([recommended_icd_code[1]])[0])
        except:
            data_top_5['pred_code2'].append(recommended_icd_code[1])
        data_top_5['score2'].append(recommended_score[1])

        try:
            data_top_5['pred_code3'].append(le.inverse_transform([recommended_icd_code[2]])[0])
        except:
            data_top_5['pred_code3'].append(recommended_icd_code[2])
        data_top_5['score3'].append(recommended_score[2])

        try:
            data_top_5['pred_code4'].append(le.inverse_transform([recommended_icd_code[3]])[0])
        except:
            data_top_5['pred_code4'].append(recommended_icd_code[3])
        data_top_5['score4'].append(recommended_score[3])

        try:
            data_top_5['pred_code5'].append(le.inverse_transform([recommended_icd_code[4]])[0])
        except:
            data_top_5['pred_code5'].append(recommended_icd_code[4])
        data_top_5['score5'].append(recommended_score[4])

    return y_pred, data_top_5
```

```python
def lgbm_custom_kfold_evaluation(X, y, le, feature, n_splits=10, top_n=10, params=params):
    X = np.array(X)
    y = np.array(y)

    kf = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=42)
    accuracy_scores, error_rates, precisions, recalls, f1_scores, index_list = [],[],[],[],[],[]

    cross_validation_results = pd.DataFrame(columns=["kfold", "accuracy", "f1_score", "error_rate", "recall", "precisions"])
    counter = 1

    for train_index, test_index in kf.split(X, y):
        index_list.append((train_index, test_index))
        X_train_fold, X_test_fold = X[train_index], X[test_index]
        y_train_fold, y_test_fold = y[train_index], y[test_index]
        y_le = list(set(y))

        model = LGBMModel('fit', X_train_fold, y_train_fold, X_test_fold, y_test_fold, params[counter-1])
        classes = model.classes_

        y_prob = model.predict_proba(X_test_fold)
        y_pred, _ = get_predictions(y_prob, classes, threshold, top_n, y_test_fold, le)

        cm = pd.DataFrame(confusion_matrix(y_test_fold, y_pred, labels=y_le))
        cm.index, cm.columns = le.inverse_transform(y_le), le.inverse_transform(y_le)

        # evaluation result per fold per class
        eval_per_fold = calculate_evaluation_metrics(Confusion_df=cm)
        eval_per_fold.to_csv(f'/content/drive/MyDrive/IF6099-TESIS/References/Bootcamp/Final Experiment/lgbm/fv_3/data_with_lk@500/data_per_fold_{feature}/eval_per_fold_{counter}.csv', index=False)

        accuracy = calculate_accuracy(cm)
        error_rate = 1 - accuracy
        precision = calculate_precision(cm)
        recall = calculate_recall(cm)
        f1_score = calculate_f1_score(precision, recall)

        cross_validation_results.loc[counter] = [int(counter), accuracy, f1_score, error_rate, recall, precision]

        accuracy_scores.append(accuracy)
        error_rates.append(error_rate)
        precisions.append(precision)
        recalls.append(recall)
        f1_scores.append(f1_score)

        counter += 1

    max_idx = np.argmax(cross_validation_results['accuracy'])
    cross_validation_results.loc[counter] = ["average", cross_validation_results["accuracy"].mean(), cross_validation_results["f1_score"].mean(), cross_validation_results["error_rate"].mean(), cross_validation_results["recall"].mean(), cross_validation_results["precisions"].mean()]
    cross_validation_results.loc[counter+1] = ["max", cross_validation_results["accuracy"][max_idx+1], cross_validation_results["f1_score"][max_idx+1], cross_validation_results["error_rate"][max_idx+1], cross_validation_results["recall"][max_idx+1], cross_validation_results["precisions"][max_idx+1]]
    cross_validation_results.to_csv(f'/content/drive/MyDrive/IF6099-TESIS/References/Bootcamp/Final Experiment/lgbm/fv_3/data_with_lk@500/cross_val_result_lgbm{feature}.csv', index=False)

    # cari fold mana yang akurasinya paling tinggi
    top_accuracy_idx = np.argmax(accuracy_scores)
    top_index_fold = index_list[top_accuracy_idx]

    # define, fit, predict, and evaluate the new model
    new_model = LGBMModel('fit', X, y, X[:11000], y[:11000], params=params[top_accuracy_idx])
    new_y_prob = new_model.predict_proba(X)
    new_classes = new_model.classes_
    new_y_pred, data_top_5 = get_predictions(new_y_prob, new_classes, threshold, top_n, y, le)

    new_cm = pd.DataFrame(confusion_matrix(y, new_y_pred))
    new_cm.index, new_cm.columns = le.inverse_transform(new_classes), le.inverse_transform(new_classes)

    eval = calculate_evaluation_metrics(Confusion_df=new_cm)
    new_cm.to_csv(f'/content/drive/MyDrive/IF6099-TESIS/References/Bootcamp/Final Experiment/lgbm/fv_3/data_with_lk@500/confusion_train_lgm{feature}.csv')
    eval.to_csv(f'/content/drive/MyDrive/IF6099-TESIS/References/Bootcamp/Final Experiment/lgbm/fv_3/data_with_lk@500/result_train_lgm{feature}.csv', index=False)

    try:
        cross_validation_results = pd.read_csv(f'/content/drive/MyDrive/IF6099-TESIS/References/Bootcamp/Final Experiment/lgbm/fv_3/data_with_lk@500/cross_val_result_lgbm{feature}.csv', index_col=False)
    except:
        print("tidak ada data, ulangi proses cross val!")

    # append data fold ke 11
    accuracy11 = calculate_accuracy(new_cm)
    error_rate11 = 1 - accuracy11
    precision11 = calculate_precision(new_cm)
    recall11 = calculate_recall(new_cm)
    f1_score11 = calculate_f1_score(precision11, recall11)
    fold11 = [11, accuracy11, f1_score11, error_rate11, recall11, precision11]

    cross_validation_results = pd.concat([cross_validation_results.iloc[:10], pd.DataFrame([fold11], columns=cross_validation_results.columns), cross_validation_results.iloc[10:]], ignore_index=True)
    cross_validation_results.to_csv(f'/content/drive/MyDrive/IF6099-TESIS/References/Bootcamp/Final Experiment/lgbm/fv_3/data_with_lk@500/cross_val_result_lgbm{feature}.csv', index=False)

    return new_model, data_top_5, new_cm, eval, cross_validation_results
```

```python
def make_dataset(data_X, data_top_5, miscellaneous):
    # Ensure we are working with a copy of the data
    data_X = data_X.copy()
    data_X['NoPendaftaran'] = miscellaneous['NoPendaftaran']
    data_X['tp_IdPasien'] = miscellaneous['tp_IdPasien']
    data_X['tp_ICD_cl1'] = miscellaneous['tp_ICD_cl1']

    # Initialize columns
    data_X["true_code"] = None
    data_X["pred_code"] = None
    data_X["pred_score"] = None

    for n in range(len(data_X)):
        top_5_pred = [data_top_5[f'pred_code{p}'][n] for p in range(1, 6)]
        top_5_score = [data_top_5[f'score{p}'][n] for p in range(1, 6)]

        if data_top_5['true_list'][n] in top_5_pred:
            top_5_pred_idx = int(top_5_pred.index(data_top_5['true_list'][n])) + 1

            # Using loc to set values
            data_X.loc[n, "true_code"] = data_top_5['true_list'][n]
            data_X.loc[n, "pred_code"] = data_top_5[f'pred_code{top_5_pred_idx}'][n]
            data_X.loc[n, "pred_score"] = data_top_5[f'score{top_5_pred_idx}'][n]

        else:
            # Using loc to set values
            data_X.loc[n, "true_code"] = data_top_5['true_list'][n]
            data_X.loc[n, "pred_code"] = data_top_5['pred_code1'][n]
            data_X.loc[n, "pred_score"] = data_top_5['score1'][n]

    data_true_train = data_X[data_X["pred_code"] == data_X["true_code"]]
    data_false_train = data_X[data_X["pred_code"] != data_X["true_code"]]

    # decode back to icd codes
    data_true_train['true_code'] = le.inverse_transform(data_true_train['true_code'])
    data_false_train['true_code'] = le.inverse_transform(data_false_train['true_code'])
    data_true_train['pred_code'] = le.inverse_transform(data_true_train['pred_code'])
    data_false_train['pred_code'] = le.inverse_transform(data_false_train['pred_code'])

    return data_true_train, data_false_train
```

```python
def calculate_dcg(scores):
    return np.sum((2**scores - 1) / np.log2(np.arange(2, scores.size + 2)))

def calculate_ndcg(Confusion_df):
    ndcgs = []
    for i, true_label in enumerate(Confusion_df.index):
        predicted_scores = Confusion_df.loc[true_label].values
        ideal_scores = np.sort(predicted_scores)[::-1]
        dcg = calculate_dcg(predicted_scores)
        idcg = calculate_dcg(ideal_scores)
        ndcg = dcg / idcg if idcg > 0 else 0
        ndcgs.append(ndcg)
    return np.mean(ndcgs)

def calculate_accuracy(Confusion_df):
    diagonal_sum = np.trace(Confusion_df.values)
    total_sum = np.sum(Confusion_df.values)
    accuracy = diagonal_sum / total_sum
    return accuracy

def calculate_precision(Confusion_df):
    precisions = []
    for label in Confusion_df.columns:
        tp = Confusion_df.at[label, label]
        column_sum = Confusion_df[label].sum()
        precision = tp / column_sum if column_sum != 0 else 0
        precisions.append(precision)
    average_precision = np.mean(precisions)
    return average_precision

def calculate_recall(Confusion_df):
    recalls = []
    for label in Confusion_df.index:
        tp = Confusion_df.at[label, label]
        row_sum = Confusion_df.loc[label].sum()
        recall = tp / row_sum if row_sum != 0 else 0
        recalls.append(recall)
    average_recall = np.mean(recalls)
    return average_recall

def calculate_f1_score(precision, recall):
    if (precision + recall) == 0:
        return 0
    return 2 * (precision * recall) / (precision + recall)

def evaluate_predictions(data, classes):
    true_list = data['true_list']
    predicted_icd_codes, predicted_highest_score = [], []

    for i in range(len(data['true_list'])):
        top_5_pred = [
            data['pred_code1'][i],
            data['pred_code2'][i],
            data['pred_code3'][i],
            data['pred_code4'][i],
            data['pred_code5'][i]
        ]
        if data['true_list'][i] in top_5_pred:
            predicted_icd_codes.append(data['true_list'][i])
            predicted_highest_score.append(data[f'score{top_5_pred.index(data["true_list"][i]) + 1}'][i])
        else:
            predicted_icd_codes.append(data['pred_code1'][i])
            predicted_highest_score.append(data['score1'][i])

    Confusion = confusion_matrix(true_list, predicted_icd_codes)
    Confusion_df = pd.DataFrame(Confusion, columns=classes, index=classes)

    data_top_1 = pd.DataFrame({
        'true_icd_code': true_list,
        'predicted_icd_codes': predicted_icd_codes,
        'predicted_highest_score': predicted_highest_score
    })

    return true_list, predicted_icd_codes, predicted_highest_score, Confusion, Confusion_df, data_top_1
```

```python
def calculate_evaluation_metrics(data=None, classes=None, Confusion_df=None):
    if not isinstance(Confusion_df, pd.DataFrame):
        true_list, predicted_icd_codes, predicted_highest_score, Confusion, Confusion_df, data_top_1 = evaluate_predictions(data, classes)

    accuracy = calculate_accuracy(Confusion_df)
    precision = calculate_precision(Confusion_df)
    recall = calculate_recall(Confusion_df)

    eval_result = pd.DataFrame(columns=['class', 'accuracy', 'hit_rate', 'error_rate', 'specificity', 'precision', 'recall', 'f1', 'type_1_error', 'type_2_error', 'ndcg'])

    for idx, label in enumerate(Confusion_df.columns):
        tp = Confusion_df.at[label, label]
        column_sum = Confusion_df[label].sum()
        row_sum = Confusion_df.loc[label].sum()
        accuracy = tp / row_sum if row_sum != 0 else 0
        error_rate = 1 - accuracy
        specificity = (column_sum - tp) / (column_sum - tp + Confusion_df.values.sum() - row_sum)
        precision = tp / column_sum if column_sum != 0 else 0
        recall = tp / row_sum if row_sum != 0 else 0
        hit_rate = recall
        f1 = calculate_f1_score(precision, recall)
        type_1_error = (Confusion_df[label].sum() - tp) / (Confusion_df[label].sum())
        type_2_error = (Confusion_df.loc[label].sum() - tp) / (Confusion_df.loc[label].sum())
        class_ndcg = calculate_ndcg(Confusion_df.loc[[label]])

        eval_result.loc[idx] = [label, accuracy, hit_rate, error_rate, specificity, precision, recall, f1, type_1_error, type_2_error, class_ndcg]

    avg_accuracy = eval_result['accuracy'].mean()
    avg_error_rate = eval_result['error_rate'].mean()
    avg_specificity = eval_result['specificity'].mean()
    avg_precision = eval_result['precision'].mean()
    avg_recall = eval_result['recall'].mean()
    avg_hit_rate = avg_recall
    avg_f1 = eval_result['f1'].mean()
    avg_type_1_error = eval_result['type_1_error'].mean()
    avg_type_2_error = eval_result['type_2_error'].mean()
    avg_ndcg = eval_result['ndcg'].mean()

    avg_row = ['average', avg_accuracy, avg_hit_rate, avg_error_rate, avg_specificity, avg_precision, avg_recall, avg_f1, avg_type_1_error, avg_type_2_error, avg_ndcg]
    eval_result.loc[len(eval_result)] = avg_row

    if not isinstance(Confusion_df, pd.DataFrame):
        return Confusion_df, eval_result, data_top_1
    else:
        return eval_result
```

```python
class CalculateTime:
    def __init__(self, features, data_type):
        self.features = features
        self.data_type = data_type

    def start(self):
        self.start_time = time.time()

    def time_taken(self):
        return (time.time() - self.start_time) / 60

    def save_time_to_csv(self, time_taken, file_name='/content/drive/MyDrive/IF6099-TESIS/References/Bootcamp/Final Experiment/lgbm/training_times.csv'):
        df = pd.DataFrame({'features': [self.features],
                           'data_type': [self.data_type],
                           'time_taken (minutes)': [time_taken]})

        df.to_csv(file_name, mode='a', header=not pd.io.common.file_exists(file_name), index=False)
```
